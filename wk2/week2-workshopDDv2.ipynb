```python
def analyze_sales(sales):
    monthly_totals = {}
    category_totals = {}
    for sale in sales:
        monthly_totals[sale["month"]] = monthly_totals.get(sale["month"], 0) + sale["amount"]
        category_totals[sale["category"]] = category_totals.get(sale["category"], 0) + sale["amount"]

    best_month = max(monthly_totals, key=monthly_totals.get)
    best_category = max(category_totals, key=category_totals.get)

    return {
        "monthly": monthly_totals,
        "category": category_totals,
        "best_month": best_month,
        "best_category": best_category
    }

results = analyze_sales(sales_data)
print("Best month:", results["best_month"], "with", results["monthly"][results["best_month"]])
print("Best category:", results["best_category"], "with", results["category"][results["best_category"]])
```
